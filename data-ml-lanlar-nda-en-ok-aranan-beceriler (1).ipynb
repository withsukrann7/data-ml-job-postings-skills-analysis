## 5. Rolleri gruplamak, alan dışı rolleri temizlemek.

Amacım veri bilimi ve makine öğrenmesiyle ilgili ilanların becerilerini incelemek. `search_position` arama sırasında kullanılan terimi gösterdiği için gerçek ilan rolü olarak kullanmıyorum; sınıflandırmada `job_title` alanını ve gerektiğinde `job_skills` içeriğini dikkate alıyorum.

İlk sınıflandırma başlık anahtar sözcükleriyle yapılır. Bu otomatik etiketleme yalnızca ilk adımdır; belirsiz eşleşmeler elle kontrol edilip düzeltilir. İlanları silmek yerine `rol_grubu` etiketiyle tutuyorum; böylece kaynak kayıtlar korunur. Data center ve data entry ilanları bu analizin kapsamı dışında bırakılıp `İlgisiz` grubuna alınır.



```python
baslik = df["job_title"].fillna("").astype(str).str.lower()

dislanan = baslik.str.contains(
    r"data\s?cent(?:er|re)|datacent(?:er|re)|data entry", regex=True,na=False
)

ilgili = baslik.str.contains(
    r"data|machine learning|\bml\b|\bai\b|analyst|analytics|scientist|"
    r"database|\bdba\b|\bsql\b|\betl\b|statistic|business intelligence|\bbi\b|mlops",
    regex=True,na=False
)

df["rol_grubu"] = np.select(
    [dislanan, ilgili],
    ["Dışlanan (data center/entry)", "Data & ML"],
    default="İlgisiz"
)

print(df["rol_grubu"].value_counts())
```

```python
print(baslik[dislanan].str.contains("entry").sum(), "data entry") #baslik burada bir metin serisi. dislanan: Aynı uzunlukta bir True/False serisi.
print(baslik[dislanan].str.contains("cent").sum(), "data center")
```

## 7. Beceri listesini hazırlama
`job_skills` sütunundaki virgülle ayrılmış becerileri analiz edebilmek için ayrı bir `skills_list` sütununa dönüştürüyorum. Ham beceri metni `job_skills` sütununda korunur. Listeyi oluştururken boşlukları temizliyor, küçük harfe çeviriyor ve aynı ilan içindeki tekrarları kaldırıyorum. Beceri bilgisi olmayan ilanlar boş liste ([]) alır.

```python
df["job_skills"] = df["job_skills"].str.strip()
```

```python
df["job_title"] = df["job_title"].str.strip()
```

```python
df["job_title"] = df["job_title"].str.lower()
```

```python
# Orijinal beceri metinlerini geri getir
df["job_skills"] = sonuc2["job_skills"]

# Virgüllere göre ayır, her becerinin başındaki/sonundaki boşlukları temizle
df["skills_list"] = df["job_skills"].apply(
    lambda x: [skill.strip() for skill in x.split(",") if skill.strip()]
    if isinstance(x, str) else []
)
```

```python
df["skills_list"] = df["skills_list"].apply(
    lambda skills: list(dict.fromkeys(
        skill.strip().lower()
        for skill in skills
        if skill.strip()
    )) if isinstance(skills, list) else []
)
```

## Kontrol: Aynı ilan birden fazla satırda olabilir mi?
Özet metinlerinin 9.661'i benzersizdi (12.217 satıra karşılık). Bazı ilanlar farklı şehir aramalarında tekrar çıkmış olabilir. Bu yüzden aynı başlık, şirket ve beceri metnine sahip satırları tek sayıp top-15 yüzdelerini karşılaştırıyorum. Sıralama değişmiyorsa sonuç tekrarlı ilanlardan etkilenmiyor demektir.

```python
tekil = df.drop_duplicates(subset=["job_title", "company", "job_skills"])
print("Tüm satırlar:", len(df), "| Tekilleştirilmiş:", len(tekil))

def yuzde_serisi(veri):
    ilan = veri.loc[
        (veri["rol_grubu"] == "Data & ML") & veri["skills_list"].apply(bool),
        ["job_link", "skills_list"]
    ]
    toplam = ilan["job_link"].nunique()
    sayi = (
        ilan.explode("skills_list")
        .groupby("skills_list")["job_link"].nunique()
        .sort_values(ascending=False)
    )
    return sayi / toplam * 100

tum_yuzde = yuzde_serisi(df)
tekil_yuzde = yuzde_serisi(tekil)

karsilastirma = pd.DataFrame({
    "tum_ilanlar_yuzde": tum_yuzde,
    "tekil_ilanlar_yuzde": tekil_yuzde
}).loc[tum_yuzde.head(15).index].round(1)

karsilastirma
```
